```python
from fredapi import Fred
import pandas as pd

fred = Fred(api_key="c424de7162b0ba5682caf74e8f6fb54e")

unemp = fred.get_series("UNRATE")
unemp = unemp.to_frame("unemployment")

ffr = fred.get_series("FEDFUNDS")
ffr = ffr.to_frame("interest_rate")

gdp = fred.get_series("A191RL1Q225SBEA")
gdp = gdp.to_frame("growth_rate")
```

```python
macro = pd.concat([unemp, ffr, gdp], axis=1)
macro = macro.sort_index()
```

```python
daily_range = pd.date_range(start=macro.index.min(), end=macro.index.max(), freq='D')

macro_daily = macro.reindex(daily_range).ffill()

macro_daily = macro_daily.reset_index().rename(columns={'index': 'date'})

macro_daily = macro_daily[(macro_daily['date'] >= "1996-01-01") & (macro_daily['date'] <= "2026-12-31")]

macro_daily
```

```python
today_date = pd.to_datetime("2026-02-17")
macro_daily = macro_daily.set_index('date')
extended_range = pd.date_range(start=macro_daily.index.min(), end=today_date, freq='D')
macro_daily = macro_daily.reindex(extended_range).ffill()
macro_daily = macro_daily.reset_index().rename(columns={'index': 'date'})
macro_daily = macro_daily[macro_daily['date'].between("1996-01-01", "2026-02-17")]
```
